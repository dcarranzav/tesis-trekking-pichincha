```python
# Validar las rutas candidatas registradas
inventario = pd.read_csv(
    proyecto / "01_documentacion" / "inventario_rutas.csv",
    encoding="utf-8",
    dtype=str,
    keep_default_na=False
)

campos_obligatorios = [
    "id_ruta",
    "nombre_ruta",
    "canton",
    "sector",
    "tipo_recorrido",
    "fuente_trazado",
    "url_fuente",
    "fecha_consulta",
    "formato_original",
    "licencia_nombre",
    "atribucion_requerida",
    "trazado_disponible",
    "elevacion_disponible",
    "calidad_preliminar",
    "decision",
]

total_vacios = int(
    inventario[campos_obligatorios]
    .apply(lambda columna: columna.str.strip().eq(""))
    .sum()
    .sum()
)

print("Número de rutas:", len(inventario))
print("Número de columnas:", len(inventario.columns))
print("Identificadores únicos:", inventario["id_ruta"].is_unique)
print("Campos obligatorios vacíos:", total_vacios)

display(
    inventario[
        [
            "id_ruta",
            "nombre_ruta",
            "canton",
            "fuente_trazado",
            "trazado_disponible",
            "calidad_preliminar",
            "decision",
        ]
    ]
)
```

```python
from pathlib import Path
import pandas as pd

# Localizar la carpeta principal del proyecto
actual = Path.cwd().resolve()

if (actual / "01_documentacion").exists():
    proyecto = actual
elif (actual.parent / "01_documentacion").exists():
    proyecto = actual.parent
else:
    raise FileNotFoundError("No se encontró la carpeta principal del proyecto")

# Cargar el inventario sin convertir textos vacíos en NaN
inventario = pd.read_csv(
    proyecto / "01_documentacion" / "inventario_rutas.csv",
    encoding="utf-8",
    dtype=str,
    keep_default_na=False
)

ids_piloto = {"R001", "R003", "R005", "R006"}
piloto = inventario[inventario["id_ruta"].isin(ids_piloto)]
dataset = pd.read_csv(
    proyecto / "07_resultados" / "tablas" / "dataset_clasificacion_rutas.csv"
)

print("Número de rutas:", len(inventario))
print("Número de columnas:", len(inventario.columns))
print("Identificadores únicos:", inventario["id_ruta"].is_unique)
print("Rutas del piloto histórico:", sorted(piloto["id_ruta"].tolist()))
print("Estados de modelado:", dataset["estado_modelado"].value_counts().to_dict())

assert len(inventario) == len(dataset)
assert inventario["id_ruta"].is_unique
assert dataset["id_ruta"].is_unique
assert set(inventario["id_ruta"]) == set(dataset["id_ruta"])
assert set(piloto["id_ruta"]) == ids_piloto

print("Inventario y dataset de clasificación coherentes")
```
